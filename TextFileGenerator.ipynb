```python
import random
from datetime import datetime, timedelta
```

```python
random.seed(42)
```

```python
places = ["the cafe", "the library", "Marco's", "the gym", "campus"]
times = ["at 1", "at 6", "after class", "tomorrow morning", "around 8"]
activities = ["lunch", "study session", "coffee", "the football match", "the group project"]
```

```python
# Each thread is a list of (speaker_role, template). Roles are mapped to real people at runtime.
threads = {
    "plan": [
        ("A", "Anyone up for {activity} {time}?"),
        ("B", "Yes! Meet at {place}?"),
        ("C", "Can't make it {time}, can we do later?"),
        ("A", "Sure, let's say {time2} then"),
        ("B", "Perfect 👍"),
    ],
    "complaint": [
        ("A", "I can't believe how much work we got for {activity}"),
        ("B", "Same 😭 I haven't even started"),
        ("A", "Let's meet at {place} {time} and get it done"),
    ],
    "share": [
        ("A", "Look at this 😂"),
        ("A", "<Media omitted>"),
        ("B", "HAHA I'm dead"),
    ],
}
```

```python
people = ["Alice", "Bob", "Charlie"]
weights = [5, 3, 2]  
```

```python
def fill(template):
    return template.format(
        place=random.choice(places),
        time=random.choice(times),
        time2=random.choice(times),
        activity=random.choice(activities),
    )
```

```python
def generate_chat(n_threads=100):
    lines = []
    current = datetime(2024, 1, 1, 9, 0)
    for _ in range(n_threads):
        current += timedelta(hours=random.randint(2, 20))
        if current.hour < 8:
            current = current.replace(hour=9)
        topic = random.choice(list(threads))

        # map roles to distinct people; first speaker is weighted
        starter = random.choices(people, weights=weights)[0]
        others = [p for p in people if p != starter]
        random.shuffle(others)
        roles = {"A": starter, "B": others[0], "C": others[1]}

        for role, template in threads[topic]:
            current += timedelta(minutes=random.randint(0, 6), seconds=random.randint(0, 59))
            lines.append(f"{current:%d/%m/%y, %H:%M} - {roles[role]}: {fill(template)}")
    return lines
```

```python
with open("sample_chat.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(generate_chat()))
```
